```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')    #This code suppresses (hides) all warning messages that Python libraries generate during execution.
```

```python
url="https://raw.githubusercontent.com/AkarshVyas/Data-Visualization-Youtube/refs/heads/main/IPL.csv"
df=pd.read_csv(url)
df.head(5)
```

```python
#BASIC INFORMATION


df.info()
```

```python
#Check the size of rows and columns of the dataset

print(f"The number of rows are {df.shape[0]} and the number of columns are {df.shape[1]}")
```

```python
#how many columns have null values in total.

df.isnull().sum()
```

```python
df.columns
```

```python
#Which team won the most matches

win_number=df['match_winner'].value_counts()
sns.barplot(y=win_number.index , x=win_number.values , palette='viridis')
plt.title("Number of wins by teams")
```

```python
#Toss Decision Trends

df['toss_decision'].value_counts()
```

```python
sns.countplot(x=df['toss_decision'] , palette='rainbow')
plt.title("Toss Decision Trends")
```

```python
#Toss winner vs Match winner

count=df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage=((count*100)/df.shape[0]).round(2)


print(f"{percentage} percentage of matches is won by the toss winning team")
```

```python
#How do teams win(runs vs wickets)?

sns.countplot(x = df['won_by'])
plt.title('Won By')
```

```python
#Most Player of the Match Awards

POM = df['player_of_the_match'].value_counts().head(10)
POM
```

```python
sns.barplot(x=POM.values , y=POM.index , palette='mako')
plt.title('Most Player of the Match Awards')
```

```python
#3 Top Scorers

high=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(3)
high
```

```python
sns.barplot(x=high.values , y=high.index)
plt.title('Top Run Scorer')
```

```python
#10 best bowling figures

df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers=df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind='barh')       #bar h(horizontal)
```

```python
#Most Matches Played by venue

venue_count=df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y=venue_count.index , x=venue_count.values , palette='rainbow' )
```

```python
#who won by the highest margin runs

df[df['won_by'] == 'Runs'].sort_values(by='margin' , ascending=False).head(1)[['match_winner' , 'margin']]
```

```python
#which player had the most individual score

df[df['highscore'] == df['highscore'].max()][['top_scorer' , 'highscore']]
```

```python
#Which bowler had the best bowling figures

df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling'  , 'best_bowling_figure']]
```
